```python
import mysql.connector
```

```python
conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="root@2026#"
)

print("Connected successfully!")

```

```python
cursor = conn.cursor()
```

```python


cursor.execute("CREATE DATABASE supply_chain_database")

print("Database created successfully!")

```

```python
cursor.execute("USE supply_chain_database")
```

```python
cursor.execute("""
CREATE TABLE suppliers (
    supplier_id INT PRIMARY KEY,
    supplier_name VARCHAR(100),
    contact_person VARCHAR(100),
    city VARCHAR(50),
    state VARCHAR(50),
    country VARCHAR(50),
    supplier_type VARCHAR(50),
    rating DECIMAL(3,2),
    status VARCHAR(20)
)
""")

conn.commit()

print("Suppliers table created successfully!")
```

```python
cursor.execute("""
CREATE TABLE products (
    product_id INT PRIMARY KEY,
    product_name VARCHAR(100),
    category VARCHAR(50),
    sub_category VARCHAR(50),
    unit_cost DECIMAL(10,2),
    selling_price DECIMAL(10,2),
    reorder_level INT,
    unit VARCHAR(20),
    status VARCHAR(20)
)
""")

conn.commit()

print("Products table created successfully!")
```

```python
cursor.execute("""
CREATE TABLE warehouses (
    warehouse_id INT PRIMARY KEY,
    warehouse_name VARCHAR(100),
    city VARCHAR(50),
    state VARCHAR(50),
    warehouse_type VARCHAR(50),
    capacity_units INT,
    manager_name VARCHAR(100),
    status VARCHAR(20)
)
""")

conn.commit()

print("Warehouses table created successfully!")
```

```python
cursor.execute("""
CREATE TABLE customers (
    customer_id INT PRIMARY KEY,
    customer_name VARCHAR(100),
    customer_type VARCHAR(30),
    city VARCHAR(50),
    state VARCHAR(50),
    country VARCHAR(50),
    registration_date DATE,
    status VARCHAR(20)
)
""")

conn.commit()

print("Customers table created successfully!")
```

```python
cursor.execute("""
CREATE TABLE purchase_orders (
    po_id INT PRIMARY KEY,
    supplier_id INT,
    warehouse_id INT,
    order_date DATE,
    expected_date DATE,
    status VARCHAR(30),
    total_amount DECIMAL(12,2),

    FOREIGN KEY (supplier_id)
        REFERENCES suppliers(supplier_id),

    FOREIGN KEY (warehouse_id)
        REFERENCES warehouses(warehouse_id)
)
""")

conn.commit()

print("Purchase orders table created successfully!")
```

```python
cursor.execute("""
CREATE TABLE purchase_order_items (
    po_item_id INT PRIMARY KEY,
    po_id INT,
    product_id INT,
    quantity INT,
    unit_price DECIMAL(10,2),
    total_amount DECIMAL(12,2),

    FOREIGN KEY (po_id)
        REFERENCES purchase_orders(po_id),

    FOREIGN KEY (product_id)
        REFERENCES products(product_id)
)
""")

conn.commit()

print("Purchase order items table created successfully!")
```

```python
cursor.execute("""
CREATE TABLE inventory (
    inventory_id INT PRIMARY KEY,
    warehouse_id INT,
    product_id INT,
    stock_date DATE,
    opening_stock INT,
    received_qty INT,
    sold_qty INT,
    closing_stock INT,
    damaged_qty INT,
    stock_value DECIMAL(12,2),

    FOREIGN KEY (warehouse_id)
        REFERENCES warehouses(warehouse_id),

    FOREIGN KEY (product_id)
        REFERENCES products(product_id)
)
""")

conn.commit()

print("Inventory table created successfully!")
```

```python
cursor.execute("""
CREATE TABLE sales_orders (
    order_id INT PRIMARY KEY,
    customer_id INT,
    warehouse_id INT,
    order_date DATE,
    required_date DATE,
    order_status VARCHAR(30),
    order_total DECIMAL(12,2),

    FOREIGN KEY (customer_id)
        REFERENCES customers(customer_id),

    FOREIGN KEY (warehouse_id)
        REFERENCES warehouses(warehouse_id)
)
""")

conn.commit()

print("Sales orders table created successfully!")
```

```python
cursor.execute("""
CREATE TABLE sales_order_items (
    order_item_id INT PRIMARY KEY,
    order_id INT,
    product_id INT,
    quantity INT,
    unit_price DECIMAL(10,2),
    discount DECIMAL(10,2),
    total_amount DECIMAL(12,2),

    FOREIGN KEY (order_id)
        REFERENCES sales_orders(order_id),

    FOREIGN KEY (product_id)
        REFERENCES products(product_id)
)
""")

conn.commit()

print("Sales order items table created successfully!")
```

```python
cursor.execute("""
CREATE TABLE shipments (
    shipment_id INT PRIMARY KEY,
    order_id INT,
    warehouse_id INT,
    shipment_date DATE,
    expected_delivery_date DATE,
    carrier_name VARCHAR(100),
    shipping_mode VARCHAR(30),
    shipment_status VARCHAR(30),
    distance_km DECIMAL(10,2),

    FOREIGN KEY (order_id)
        REFERENCES sales_orders(order_id),

    FOREIGN KEY (warehouse_id)
        REFERENCES warehouses(warehouse_id)
)
""")

conn.commit()

print("Shipments table created successfully!")
```

```python
cursor.execute("""
CREATE TABLE deliveries (
    delivery_id INT PRIMARY KEY,
    shipment_id INT,
    delivery_date DATE,
    delivery_status VARCHAR(30),
    delivery_delay_days INT,
    delivery_rating DECIMAL(3,2),
    delivery_notes VARCHAR(255),

    FOREIGN KEY (shipment_id)
        REFERENCES shipments(shipment_id)
)
""")

conn.commit()

print("Deliveries table created successfully!")
```

```python
cursor.execute("""
CREATE TABLE transportation (
    transport_id INT PRIMARY KEY,
    shipment_id INT,
    carrier_name VARCHAR(100),
    vehicle_type VARCHAR(50),
    distance_km DECIMAL(10,2),
    fuel_cost DECIMAL(10,2),
    freight_cost DECIMAL(10,2),
    handling_cost DECIMAL(10,2),
    total_transport_cost DECIMAL(12,2),

    FOREIGN KEY (shipment_id)
        REFERENCES shipments(shipment_id)
)
""")

conn.commit()

print("Transportation table created successfully!")
```

```python
cursor.execute("""
SELECT TABLE_NAME
FROM INFORMATION_SCHEMA.TABLES
WHERE TABLE_SCHEMA = 'supply_chain_database'
  AND TABLE_TYPE = 'BASE TABLE'
ORDER BY TABLE_NAME
""")

tables = cursor.fetchall()

for table in tables:
    print(table[0])
```
