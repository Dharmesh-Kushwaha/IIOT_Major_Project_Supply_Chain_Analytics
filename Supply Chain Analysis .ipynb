##### Part -> 1 Importing Library & Reading the Dataframes

```python
#Importing Necessery Library
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_excel("supply chainupdated data.xlsx",sheet_name=None)
```

```python
# Access Individual DataFrames
warehouses = data["warehouses"]
transportation = data["transportation"]
suppliers = data["suppliers"]
shipments = data["shipments"]
sales_orders = data["sales_orders"]
sales_order_items = data["sales_order_items"]
purchase_orders = data["purchase_orders"]
purchase_order_items = data["purchase_order_items"]
products = data["products"]
inventory = data["inventory"]
deliveries = data["deliveries"]
customers = data["customers"]
```

##### Part - 2 Data Cleaning

```python
#Display the Shape of each dataframe
for name, df in data.items():
    print(name, df.shape)
```

```python
# Reading the 5 rows for every dataframe 
for name, df in data.items():
    print("\nDataset:", name)
    display(df.head())
```

```python
#Statistical description of the data in a DataFrame
for name, df in data.items():
    print("\nDataset:", name)
    display(df.describe())
```

```python
# Displays a summary of the dataFrames including column names, non-null counts, and data types.
for name, df in data.items():
    print("\nDataset:", name)
    df.info()
```

```python
#Checking duplicate values
for name, df in data.items():
    print("\nDataset:", name)
    print("Duplicate Rows:", df.duplicated().sum())
```

```python
#Checking null values
for name, df in data.items():
    print("\nDataset:", name)
    print(df.isnull().sum())
```

##### Part -> 6 Exploratory Data Analysis

## Warehouse Analysis

```python
type_data = warehouses['warehouse_type'].value_counts()
```

```python
# Warehouse Type Contribution
plt.figure(figsize=(7, 7))
plt.pie(type_data,labels=type_data.index,autopct='%1.1f%%')
plt.title("Warehouse Type Contribution")
plt.show()
```

```python
# Warehouse capacity
capacity = warehouses.groupby("warehouse_name")["capacity_units"].sum().sort_values()
```

```python
#Capacity Units by Warehouse
plt.figure(figsize=(10, 6))
plt.bar(capacity.index,capacity.values)
plt.title("Capacity Units by Warehouse", fontsize=14)
plt.xlabel("Warehouse")
plt.ylabel("Capacity Units")
plt.xticks(rotation=45)
plt.show()
```

```python
capacity_data = warehouses.groupby('warehouse_type')['capacity_units'].sum()
capacity_data
```

```python
# Warehouse Capacity by Type
plt.figure(figsize=(9, 5))
plt.bar(capacity_data.index, capacity_data.values, color='orange')
plt.title("Total Capacity by Warehouse Type")
plt.xlabel("Warehouse Type")
plt.ylabel("Capacity Units")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
# Warehouse Capacity Comparison
plt.figure(figsize=(11, 5))
plt.bar(warehouses['warehouse_name'],warehouses['capacity_units'],color='steelblue')
plt.title("Warehouse Capacity Comparison")
plt.xlabel("Warehouse Name")
plt.ylabel("Capacity Units")
plt.xticks(rotation=45)
plt.show()
```

```python
# Warehouse Capacity by City
city_data = warehouses.groupby('city')['capacity_units'].sum()
```

```python
plt.figure(figsize=(10, 5))
sns.barplot(x=city_data.values, y=city_data.index)
plt.title("Warehouse Capacity by City")
plt.xlabel("Capacity Units")
plt.ylabel("City")
plt.show()
```

```python
# Merge inventory with warehouses
inventory_warehouse = pd.merge(inventory,warehouses,on="warehouse_id",how="left")
```

```python
# Calculate closing stock by warehouse
stock_by_warehouse = inventory_warehouse.groupby("warehouse_name")["closing_stock"].sum().sort_values()
```

```python
#CLosing Stock by Warehouse
plt.figure(figsize=(10, 6))
plt.barh(stock_by_warehouse.index,stock_by_warehouse.values)
plt.title("Closing Stock by Warehouse", fontsize=14)
plt.xlabel("Total Closing Stock")
plt.ylabel("Warehouse")
plt.show()
```

```python
# Calculate damaged quantity by warehouse
damaged_stock = inventory_warehouse.groupby("warehouse_name")["damaged_qty"].sum().sort_values()
```

```python
#Damaged Quantity by Warehouse
plt.figure(figsize=(10, 6))
plt.barh(damaged_stock.index,damaged_stock.values)
plt.title("Damaged Quantity by Warehouse", fontsize=14)
plt.xlabel("Damaged Quantity")
plt.ylabel("Warehouse")
plt.show()
```

### Product And Category Analysis

```python
category_count = products["category"].value_counts()
category_count
```

```python
#Category Wise Product Distribution
plt.figure(figsize=(9, 6))
plt.pie(category_count, labels=category_count.index, autopct="%1.1f%%")
plt.title("Category Wise Product Distribution")
plt.show()
```

```python
# Merge sales items with products
sales_product = pd.merge(sales_order_items,products,on="product_id",how="left")
```

```python
# Calculate sales amount by category
sales_category = sales_product.groupby("category")["total_amount"].sum().sort_values()
```

```python
# Sales by Category
plt.figure(figsize=(9, 6))
plt.bar(sales_category.index,sales_category.values)
plt.title("Sales Amount by Category", fontsize=14)
plt.xlabel("Product Category")
plt.ylabel("Total Sales Amount")
plt.show()
```

```python
#Avg selling Price by Category
category_price = products.groupby("category")["selling_price"].mean()
category_price
```

```python
#Average Selling Price by Category
sns.barplot(x=category_price.index, y=category_price.values)
plt.title("Average Selling Price by Category")
plt.xlabel("Category")
plt.ylabel("Average Selling Price")
plt.show()
```

```python
#Unit Cost vs Selling Price
sns.scatterplot(data=products, x="unit_cost", y="selling_price", hue="category")
plt.title("Unit Cost vs Selling Price")
plt.xlabel("Unit Cost")
plt.ylabel("Selling Price")
plt.show()
```

```python
products
```

```python
# Calculate profit per unit
products["profit_per_unit"] = (products["selling_price"] - products["unit_cost"])
```

```python
#Top 10 Profitable Products
top_products = products.sort_values("profit_per_unit", ascending=False).head(10)
sns.barplot(x=top_products["profit_per_unit"], y=top_products["product_name"])
plt.title("Top 10 Profitable Products")
plt.xlabel("Profit Per Unit")
plt.ylabel("Product Name")
plt.show()
```

```python
# Convert order_date into datetime
sales_orders["order_date"] = pd.to_datetime(sales_orders["order_date"])
```

```python
# Group orders by month
monthly_orders = (sales_orders.set_index("order_date").resample("ME")["order_id"].count())
```

```python
# Monthly Order Trend
plt.figure(figsize=(12, 6))
plt.plot(monthly_orders.index,monthly_orders.values,marker="o",linewidth=2)
plt.title("Monthly Order Trend", fontsize=14)
plt.xlabel("Month")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)
plt.show()
```

```python
fig, ax = plt.subplots(3, 3, figsize=(20, 15))

fig.suptitle("SUPPLY CHAIN ANALYTICS DASHBOARD",fontsize=20, fontweight="bold", y=0.98)

# 1. Delivery Status 
ax[0,0].pie(delivery_count, labels=delivery_count.index, autopct="%1.1f%%",textprops={"fontsize": 9})
ax[0,0].set_title("Delivery Status", fontsize=12, pad=12)

# 2. Customer Type
ax[0,1].pie(customer_count, labels=customer_count.index, autopct="%1.1f%%",textprops={"fontsize": 9})
ax[0,1].set_title("Customer Type", fontsize=12, pad=12)

# 3. Sales by Product Category
ax[0,2].barh( sales_category.index,sales_category.values)
ax[0,2].set_title("Sales by Product Category",fontsize=12,pad=12)
ax[0,2].tick_params(axis="y", labelsize=8)

# 4. Closing Stock by Warehouse
ax[1,0].bar(stock.index, stock.values)
ax[1,0].set_title("Closing Stock by Warehouse", fontsize=12, pad=12)
ax[1,0].tick_params(axis="x", rotation=45, labelsize=8)

# 5. Transportation Cost by Vehicle
ax[1,1].bar(transport.index, transport.values)
ax[1,1].set_title("Transportation Cost by Vehicle", fontsize=12, pad=12)
ax[1,1].tick_params(axis="x", rotation=30, labelsize=8)

# 6. Damaged Quantity by Warehouse
ax[1,2].bar(damage.index.astype(str), damage.values)
ax[1,2].set_title("Damaged Quantity by Warehouse", fontsize=12, pad=12)
ax[1,2].tick_params(axis="x", rotation=45, labelsize=8)

# 7. Distance vs Transportation Cost
ax[2,0].scatter(transportation["distance_km"],transportation["total_transport_cost"])
ax[2,0].set_title("Distance vs Transportation Cost", fontsize=12, pad=12)
ax[2,0].set_xlabel("Distance (KM)", fontsize=9)
ax[2,0].set_ylabel("Transport Cost", fontsize=9)

# 8. Closing Stock vs Damaged Quantity
ax[2,1].scatter(inventory["closing_stock"],inventory["damaged_qty"])
ax[2,1].set_title("Closing Stock vs Damaged Quantity", fontsize=12, pad=12)
ax[2,1].set_xlabel("Closing Stock", fontsize=9)
ax[2,1].set_ylabel("Damaged Quantity", fontsize=9)

# 9. Monthly Order Trend
ax[2,2].plot(monthly_orders.index, monthly_orders.values, marker="o")
ax[2,2].set_title("Monthly Order Trend", fontsize=12, pad=12)
ax[2,2].tick_params(axis="x", rotation=45, labelsize=8)

# Spacing
plt.tight_layout(rect=[0, 0, 1, 0.95], h_pad=3, w_pad=2)

plt.show()
```
